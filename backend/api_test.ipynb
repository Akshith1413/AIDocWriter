# Aureview API LLM Test

This notebook demonstrates direct access to the `AgentLLM` class and structured output tool calling.

```
import asyncio
from app.llm import AgentLLM, ModelSelection
from app.templates import get_template

template = get_template("prd")
client = AgentLLM(ModelSelection("demo", "studio-demo"))
title = "Customer Onboarding Portal"
notes = "Need an onboarding portal. It should be fast and secure. Include audit logs."

async def test_generation():
    draft = await client.draft(template, title, notes, 1)
    print("DRAFT GENERATED:")
    print(draft[:200] + "...")
    return draft

draft = await test_generation()
```

```
async def test_review(draft):
    review = await client.review(template, notes, draft)
    print("\nREVIEW RESULTS:")
    print("Status:", review.status)
    print("Score:", review.score)
    print("Findings:", len(review.findings))
    if review.findings:
        print("First finding:", review.findings[0].issue)

await test_review(draft)
```
